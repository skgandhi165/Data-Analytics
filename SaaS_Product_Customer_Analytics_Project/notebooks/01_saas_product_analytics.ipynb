# SaaS Product & Customer Analytics
MRR → retention → engagement → revenue risk → customer health

```python
import pandas as pd
from pathlib import Path
ROOT=Path('..')
a=pd.read_csv(ROOT/'data/raw/saas_accounts_seed.csv',parse_dates=['signup_date','churn_date'])
u=pd.read_csv(ROOT/'data/raw/saas_feature_usage_seed.csv')
s=pd.read_csv(ROOT/'data/raw/saas_support_seed.csv')
```

```python
a['churn_flag']=a.churn_date.notna()
print('Churn rate:', a.churn_flag.mean())
print('Revenue at risk:', a.loc[a.churn_flag,'mrr'].sum())
```

```python
plan=a.groupby('plan').agg(accounts=('account_id','count'),churn_rate=('churn_flag','mean'),mrr=('mrr','sum')).sort_values('churn_rate',ascending=False)
plan
```

```python
usage=u.groupby('account_id').agg(feature_uses=('usage_count','sum'))
support=s.groupby('account_id').agg(tickets=('ticket_id','count'),avg_satisfaction=('satisfaction_score','mean'))
health=a.join(usage).join(support).fillna(0)
health.groupby('churn_flag')[['feature_uses','tickets','avg_satisfaction','mrr']].mean()
```
