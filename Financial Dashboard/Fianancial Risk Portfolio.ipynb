```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
# Import tools from scikit-learn for machine learning

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    roc_auc_score
)

# Set a random seed so the simulated dataset is reproducible
np.random.seed(42)
```

```python
# Define the number of simulated customer accounts in our portfolio
num_accounts = 10000

customer_ids = [f"C{i:05d}" for i in range(1, num_accounts + 1)]

print(customer_ids[:10])
```

```python
# Generate a simulated age for each customer
ages = np.random.randint(18, 76, size=num_accounts)

print(ages[:10])
```

```python
# Generate a simulated annual income for each customer
incomes = np.random.randint(30000, 150001, size=num_accounts)

print(incomes[:10])
```

```python
# Generate a simulated credit score for each customer
credit_scores = np.random.randint(300, 851, size=num_accounts)

print(credit_scores[:10])
```

```python
# Generate simulated years of employment for each customer
employment_years = np.random.randint(0, 41, size=num_accounts)

print(employment_years[:10])
```

```python
# Generate simulated account tenure in years
account_tenure = np.random.randint(0, 21, size=num_accounts)

print(account_tenure[:10])
```

```python
# Generate a simulated outstanding loan amount for each customer
loan_amounts = np.random.randint(1000, 100001, size=num_accounts)

print(loan_amounts[:10])
```

```python
# Generate a simulated annual interest rate for each customer's loan
interest_rates = np.round(
    np.random.uniform(3, 15, size=num_accounts),
    2
)

print(interest_rates[:10])
```

```python
# Generate simulated credit utilization percentages
credit_utilization = np.round(
    np.random.uniform(5, 95, size=num_accounts),
    2
)

print(credit_utilization[:10])
```

```python
# Generate the number of late payments for each customer
late_payments = np.random.randint(0, 11, size=num_accounts)

print(late_payments[:10])
```

```python
# Generate the number of missed payments during the last 6 months
missed_payments_6m = np.random.randint(0, 7, size=num_accounts)

print(missed_payments_6m[:10])
```

```python
# Generate the average number of monthly transactions for each customer
transaction_count = np.random.randint(5, 151, size=num_accounts)

print(transaction_count[:10])
```

```python
# Generate simulated average monthly spending for each customer
avg_monthly_spend = np.round(
    np.random.uniform(500, 10000, size=num_accounts),
    2
)

print(avg_monthly_spend[:10])
```

```python
# Generate a simulated monthly debt payment for each customer
monthly_debt_payment = np.round(
    np.random.uniform(200, 5000, size=num_accounts),
    2
)

print(monthly_debt_payment[:10])
```

```python
# Generate a realistic target DTI for each customer
# This represents the proportion of monthly income committed to debt payments.
dti = np.round(
    np.random.uniform(5, 60, size=num_accounts),
    2
)

# Calculate monthly income from annual income
monthly_income = incomes / 12

# Calculate monthly debt payment from income and DTI
monthly_debt_payment = np.round(
    monthly_income * (dti / 100),
    2
)

print("DTI:", dti[:10])
print("Monthly debt payment:", monthly_debt_payment[:10])
```

```python
# Generate a realistic loan-to-income ratio for each customer
loan_to_income = np.random.uniform(0.10, 1.50, size=num_accounts)

# Calculate loan amount based on annual income
loan_amounts = np.round(
    incomes * loan_to_income,
    -2
)

print(loan_amounts[:10])
```

```python
# Generate recent missed payments based on the customer's overall late-payment history
missed_payments_6m = np.zeros(num_accounts, dtype=int)

for i in range(num_accounts):
    if late_payments[i] == 0:
        missed_payments_6m[i] = 0
    else:
        # Recent missed payments cannot exceed the customer's total late payments
        missed_payments_6m[i] = np.random.randint(
            0,
            min(late_payments[i], 3) + 1
        )

print(missed_payments_6m[:10])
```

```python
# Generate credit utilization with some relationship to the customer's DTI
base_utilization = np.random.uniform(10, 70, size=num_accounts)

credit_utilization = np.round(
    base_utilization + (dti * 0.4),
    2
)

# Keep utilization within a realistic 5%–95% range
credit_utilization = np.clip(
    credit_utilization,
    5,
    95
)

print(credit_utilization[:10])
```

```python
# Adjust credit scores based on financial and payment behavior
credit_scores = (
    850
    - (dti * 2.5)
    - (credit_utilization * 1.2)
    - (late_payments * 8)
    + np.random.normal(0, 35, size=num_accounts)
)

# Keep credit scores within the standard 300–850 range
credit_scores = np.clip(
    credit_scores,
    300,
    850
).round().astype(int)

print(credit_scores[:10])
```

```python
# Create a payment history score based on late and recent missed payments
payment_history_score = (
    100
    - (late_payments * 5)
    - (missed_payments_6m * 10)
)

# Keep the score between 0 and 100
payment_history_score = np.clip(
    payment_history_score,
    0,
    100
).round().astype(int)

print(payment_history_score[:10])
```

```python
# Calculate a default-risk score based on customer financial behavior
default_risk_score = (
    (dti * 0.04)
    + (credit_utilization * 0.02)
    + (late_payments * 0.08)
    + (missed_payments_6m * 0.15)
    + ((700 - credit_scores) * 0.003)
)

# Convert the risk score into a probability between 0 and 1
default_probability = 1 / (1 + np.exp(-default_risk_score + 3))

# Generate the final default outcome
default = np.random.binomial(
    1,
    default_probability,
    size=num_accounts
)

print(default[:20])
```

```python
# Count the number of customers who defaulted
num_defaults = default.sum()

# Calculate the overall default rate
default_rate = default.mean() * 100

print("Number of defaults:", num_defaults)
print("Default rate:", round(default_rate, 2), "%")
```

```python
# Combine all customer attributes into one Pandas DataFrame
customer_data = pd.DataFrame({
    "customer_id": customer_ids,
    "age": ages,
    "annual_income": incomes,
    "credit_score": credit_scores,
    "employment_years": employment_years,
    "account_tenure": account_tenure,
    "loan_amount": loan_amounts,
    "interest_rate": interest_rates,
    "credit_utilization": credit_utilization,
    "late_payments": late_payments,
    "missed_payments_6m": missed_payments_6m,
    "transaction_count": transaction_count,
    "avg_monthly_spend": avg_monthly_spend,
    "monthly_debt_payment": monthly_debt_payment,
    "dti": dti,
    "payment_history_score": payment_history_score,
    "default": default
})

print(customer_data.head())
```

```python
# Create credit score segments for portfolio analysis
credit_score_bins = [299, 579, 669, 739, 799, 850]
credit_score_labels = [
    "Poor",
    "Fair",
    "Good",
    "Very Good",
    "Excellent"
]

customer_data["credit_score_segment"] = pd.cut(
    customer_data["credit_score"],
    bins=credit_score_bins,
    labels=credit_score_labels
)

# Calculate default rate for each credit score segment
default_by_credit_score = (
    customer_data
    .groupby("credit_score_segment", observed=True)["default"]
    .mean()
    .mul(100)
    .round(2)
)

print(default_by_credit_score)
```

```python
# Create DTI segments for portfolio risk analysis
dti_bins = [0, 20, 40, 60]
dti_labels = [
    "Low DTI",
    "Moderate DTI",
    "High DTI"
]

customer_data["dti_segment"] = pd.cut(
    customer_data["dti"],
    bins=dti_bins,
    labels=dti_labels,
    include_lowest=True
)

# Calculate default rate for each DTI segment
default_by_dti = (
    customer_data
    .groupby("dti_segment", observed=True)["default"]
    .mean()
    .mul(100)
    .round(2)
)

print(default_by_dti)
```

```python
# Create credit utilization segments for risk analysis
utilization_bins = [0, 30, 70, 95]
utilization_labels = [
    "Low Utilization",
    "Moderate Utilization",
    "High Utilization"
]

customer_data["utilization_segment"] = pd.cut(
    customer_data["credit_utilization"],
    bins=utilization_bins,
    labels=utilization_labels,
    include_lowest=True
)

# Calculate default rate for each utilization segment
default_by_utilization = (
    customer_data
    .groupby("utilization_segment", observed=True)["default"]
    .mean()
    .mul(100)
    .round(2)
)

print(default_by_utilization)
```

```python
# Create payment history segments for risk analysis
payment_history_bins = [0, 59, 79, 100]
payment_history_labels = [
    "Weak",
    "Moderate",
    "Strong"
]

customer_data["payment_history_segment"] = pd.cut(
    customer_data["payment_history_score"],
    bins=payment_history_bins,
    labels=payment_history_labels,
    include_lowest=True
)

# Calculate default rate for each payment history segment
default_by_payment_history = (
    customer_data
    .groupby("payment_history_segment", observed=True)["default"]
    .mean()
    .mul(100)
    .round(2)
)

print(default_by_payment_history)
```

```python
# Calculate default rate based on recent missed payments
default_by_missed_payments = (
    customer_data
    .groupby("missed_payments_6m", observed=True)["default"]
    .mean()
    .mul(100)
    .round(2)
)

print(default_by_missed_payments)
```

```python
# Create income segments for portfolio analysis
income_bins = [0, 50000, 80000, 120000, float("inf")]
income_labels = [
    "Low Income",
    "Middle Income",
    "Upper-Middle Income",
    "High Income"
]

customer_data["income_segment"] = pd.cut(
    customer_data["annual_income"],
    bins=income_bins,
    labels=income_labels,
    include_lowest=True
)

# Calculate default rate for each income segment
default_by_income = (
    customer_data
    .groupby("income_segment", observed=True)["default"]
    .mean()
    .mul(100)
    .round(2)
)

print(default_by_income)
```

```python
# Create age segments for customer analysis
age_bins = [17, 25, 35, 50, 65, 75]
age_labels = [
    "18-25",
    "26-35",
    "36-50",
    "51-65",
    "66-75"
]

customer_data["age_segment"] = pd.cut(
    customer_data["age"],
    bins=age_bins,
    labels=age_labels,
    include_lowest=True
)

# Calculate default rate for each age segment
default_by_age = (
    customer_data
    .groupby("age_segment", observed=True)["default"]
    .mean()
    .mul(100)
    .round(2)
)

print(default_by_age)
```

```python
# Calculate correlations between key risk factors and default
risk_correlation = customer_data[
    [
        "credit_score",
        "dti",
        "credit_utilization",
        "late_payments",
        "missed_payments_6m",
        "payment_history_score",
        "default"
    ]
].corr()["default"].sort_values()

print(risk_correlation)
```

```python
# Count the number of risk indicators present for each customer
customer_data["risk_indicator_count"] = (
    (customer_data["credit_score"] < 600).astype(int)
    + (customer_data["dti"] > 40).astype(int)
    + (customer_data["credit_utilization"] > 70).astype(int)
    + (customer_data["late_payments"] >= 3).astype(int)
    + (customer_data["missed_payments_6m"] >= 1).astype(int)
)

print(customer_data["risk_indicator_count"].value_counts().sort_index())
```

```python
# Classify customers into initial risk tiers based on the number of risk indicators
def assign_risk_tier(risk_count):
    if risk_count <= 1:
        return "Low Risk"
    elif risk_count <= 3:
        return "Medium Risk"
    else:
        return "High Risk"


customer_data["risk_tier"] = (
    customer_data["risk_indicator_count"]
    .apply(assign_risk_tier)
)

print(customer_data["risk_tier"].value_counts())
```

```python
# Calculate default rate for each rule-based risk tier
default_by_risk_tier = (
    customer_data
    .groupby("risk_tier", observed=True)["default"]
    .agg(["count", "sum", "mean"])
)

# Convert default rate from a decimal to a percentage
default_by_risk_tier["default_rate"] = (
    default_by_risk_tier["mean"] * 100
).round(2)

print(default_by_risk_tier)
```

```python
# Define the features used to predict customer default
feature_columns = [
    "age",
    "annual_income",
    "credit_score",
    "employment_years",
    "account_tenure",
    "loan_amount",
    "interest_rate",
    "credit_utilization",
    "late_payments",
    "missed_payments_6m",
    "transaction_count",
    "avg_monthly_spend",
    "monthly_debt_payment",
    "dti",
    "payment_history_score"
]

# Create the feature dataset
X = customer_data[feature_columns]

# Create the target variable
y = customer_data["default"]

print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python
# Import the function used to split our dataset
from sklearn.model_selection import train_test_split

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)
print("Training target:", y_train.shape)
print("Testing target:", y_test.shape)
```

```python
# Create a scaler to standardize our numerical features
scaler = StandardScaler()

# Learn the scaling parameters from the training data
X_train_scaled = scaler.fit_transform(X_train)

# Apply the same scaling to the test data
X_test_scaled = scaler.transform(X_test)

print("Training data shape:", X_train_scaled.shape)
print("Testing data shape:", X_test_scaled.shape)
```

```python
# Create the Logistic Regression classification model
logistic_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

# Train the model using the training data
logistic_model.fit(
    X_train_scaled,
    y_train
)

print("Logistic Regression model trained successfully!")
```

```python
# Predict whether each customer in the test set will default
y_pred = logistic_model.predict(X_test_scaled)

# Generate the predicted probability of default
y_pred_probability = logistic_model.predict_proba(X_test_scaled)[:, 1]

print("First 10 predicted outcomes:")
print(y_pred[:10])

print("\nFirst 10 predicted default probabilities:")
print(y_pred_probability[:10])
```

```python
# Calculate the model's classification accuracy
accuracy = accuracy_score(y_test, y_pred)

print("Model accuracy:", round(accuracy * 100, 2), "%")
```

```python
# Create a confusion matrix to evaluate model predictions
cm = confusion_matrix(y_test, y_pred)

print("Confusion Matrix:")
print(cm)
```

```python
# Calculate precision, recall, and F1 score
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

print("Precision:", round(precision * 100, 2), "%")
print("Recall:", round(recall * 100, 2), "%")
print("F1 Score:", round(f1 * 100, 2), "%")
```

```python
# Calculate the ROC-AUC score using predicted default probabilities
roc_auc = roc_auc_score(
    y_test,
    y_pred_probability
)

print("ROC-AUC:", round(roc_auc, 3))
```

```python
# Count how many test customers the model is currently flagging as potential defaults
predicted_defaults = y_pred.sum()
total_test_customers = len(y_pred)

predicted_default_rate = (
    predicted_defaults / total_test_customers
) * 100

print("Test customers:", total_test_customers)
print("Predicted defaults:", predicted_defaults)
print(
    "Percentage flagged for review:",
    round(predicted_default_rate, 2),
    "%"
)
```

```python
# Apply a more cautious 40% probability threshold
threshold = 0.40

y_pred_40 = (
    y_pred_probability >= threshold
).astype(int)

# Calculate performance at the new threshold
accuracy_40 = accuracy_score(y_test, y_pred_40)
precision_40 = precision_score(y_test, y_pred_40)
recall_40 = recall_score(y_test, y_pred_40)
f1_40 = f1_score(y_test, y_pred_40)

print("Threshold:", threshold)
print("Accuracy:", round(accuracy_40 * 100, 2), "%")
print("Precision:", round(precision_40 * 100, 2), "%")
print("Recall:", round(recall_40 * 100, 2), "%")
print("F1 Score:", round(f1_40 * 100, 2), "%")
```

```python
# Compare model performance at different probability thresholds
thresholds = [0.30, 0.40, 0.50, 0.60, 0.70]

for threshold in thresholds:
    predictions = (
        y_pred_probability >= threshold
    ).astype(int)

    accuracy_value = accuracy_score(y_test, predictions)
    precision_value = precision_score(y_test, predictions, zero_division=0)
    recall_value = recall_score(y_test, predictions, zero_division=0)

    print(
        f"Threshold: {threshold:.0%} | "
        f"Accuracy: {accuracy_value:.2%} | "
        f"Precision: {precision_value:.2%} | "
        f"Recall: {recall_value:.2%}"
    )
```

```python
# Calculate the number of accounts flagged for manual review
# at each probability threshold
total_accounts_for_review = len(y_test)

for threshold in thresholds:
    predictions = (
        y_pred_probability >= threshold
    ).astype(int)

    flagged_accounts = predictions.sum()

    review_percentage = (
        flagged_accounts / total_accounts_for_review
    ) * 100

    workload_reduction = 100 - review_percentage

    print(
        f"Threshold: {threshold:.0%} | "
        f"Flagged: {flagged_accounts} | "
        f"Review volume: {review_percentage:.2f}% | "
        f"Workload reduction: {workload_reduction:.2f}%"
    )
```

```python
# Calculate portfolio exposure and account counts by ML risk tier
risk_tier_summary = (
    final_customer_data
    .groupby("ml_risk_tier", observed=True)
    .agg(
        accounts=("customer_id", "count"),
        total_exposure=("loan_amount", "sum"),
        average_exposure=("loan_amount", "mean"),
        actual_defaults=("default", "sum")
    )
)

# Calculate the observed default rate
risk_tier_summary["default_rate"] = (
    risk_tier_summary["actual_defaults"]
    / risk_tier_summary["accounts"]
    * 100
).round(2)

# Calculate each tier's share of total portfolio exposure
total_exposure = final_customer_data["loan_amount"].sum()

risk_tier_summary["exposure_share"] = (
    risk_tier_summary["total_exposure"]
    / total_exposure
    * 100
).round(2)

print(risk_tier_summary)
```

```python
# Create a clean summary of portfolio risk by risk tier
portfolio_summary = (
    final_customer_data
    .groupby("ml_risk_tier", observed=True)
    .agg(
        account_count=("customer_id", "count"),
        total_exposure=("loan_amount", "sum"),
        default_count=("default", "sum")
    )
    .reset_index()
)

# Calculate default rate
portfolio_summary["default_rate"] = (
    portfolio_summary["default_count"]
    / portfolio_summary["account_count"]
    * 100
).round(2)

# Calculate exposure share
portfolio_summary["exposure_share"] = (
    portfolio_summary["total_exposure"]
    / portfolio_summary["total_exposure"].sum()
    * 100
).round(2)

print(portfolio_summary)
```
